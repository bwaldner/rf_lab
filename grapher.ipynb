```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv('example_data_1_PHYS_433_533.csv')
freq = df["f_res"]
pow = df["max_power_dBm"]
mask1 = freq >4000
freq_filter = freq[mask1]

```

```python
plt.plot(freq,pow)
plt.title("P vs F")
plt.xlabel("h")
plt.ylabel("HH)")
plt.show()
```

```python
def linear_fit(m,x,b):
    return m*x + b

popt, pcov = curve_fit(linear_fit, freq, pow, p0=[-1,4e-5])
print(popt)
print(pcov)

plt.title("Pow vs. Freq filt w lin fit")
plt.scatter(freq,pow)
plt.plot(freq_filter,linear_fit(freq_filter, popt[0],popt[1]),color='red')
```
